# Baseline for Model 2 — untuned `t5-small` on the same held-out test

Model 2 is a full fine-tune of the plain `t5-small` checkpoint (`train_conditions_model.py`
`--base-model t5-small`). Scoring that untuned checkpoint on the same test set gives the
reference point the conditions task currently lacks: it separates what fine-tuning
contributes from what the base language model already does.

Same test set and decoding parameters as the two fine-tuned configurations:
`v2_ord_conditions_test_clean.jsonl` (5,687 rows), 10 beams, top-1/3/5.

Input: dataset `kuzmenkooleh/retro-planner-ord-conditions-test-clean`.
